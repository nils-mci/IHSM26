### Variables

As an alternative to direct output we can store output in so called *variables*. To assign a value to a variable use the assignment operator "<-"

```r
x <- 3*3
```

or simply use the euqation sign "="

```r
x = 3*3
```

Both operators assign a value to a variable from right to left. It doesn't matter which one you use, they're functionally identical. However, it is highly recommended not to mix both spellings. As, to my experience, most IHSM students prefer "=", the simple equation sign is used in class throughout.

Values assigned to variables are not printed. In order to see the value, we have to print the content of the variable:

```r
x
```

R Variables are simple placeholders that can be used in further calculations:

```r
x+3
```

The above statement ist equal to (3*3)+3:

```r
(3*3)+3
```

Note that each cell prints its last output only:

```r
3*4
1+2
```

but

```r
1+2
3*4
```

We can calcualte with as many variables we want:

```r
y=12
z=2
(x+y)*z
```

### Built-in R Functions

R provides different functions that can be applied to text, numbers, or placeholder variables. The following statement

```r
x + y + z
```

could make use of a function called *sum()* which simply summarizes all values defined in the round brackets (), e.g.

```r
sum(x, y, z)
```

### Collections

This is working code, but not at all elegant. In data analysis, we mostly deal with variables that represent collections (vectors, sets, lists of data). In R, the most convenient and most widely used way to create a collection is by applying the function c().


```r
v = c(x, y, z)
```

The new collection holds three values, namely:

```r
v
```

You may think of this collection as a variable in its core statistical meaning, i.e. as values of three subjects, such as achieved points in an exam (max. 12 points) by three different students. We can now use this collection as an argument to different functions. 

In total, the students' points sum up to 

```r
sum(v)
```

while students achieved 

```r
mean(v)
```

points on average.

### Indexing

While round brackets () are used in calculations and to enclose function arguments, squared brackets [] are used to define  indices, i.e., positions in a list: 

The second student in our list reached

```r
v[2]
```

points, while the third has failed with only 

```r
v[3]
```

points. The points of both above mentioned students are 

```r
v[c(2,3)]
```

This is equal to the points of all except the first student: 

```r
v[-1]
```

### Logical vectors

We can access specific values by their characteristics, too. 

To pass the exam, students have to achieve at least 7 points.  Who has passed? 

```r
v >= 7
```

The first two students passed (v >= 7 is TRUE), student no. 3 failed (v >= 7 is FALSE). In R, this is called a *logical* vector.

We can use logical vectors as indices in order to create subsets of our data, because R returns values of collections with TRUE index while values with FALSE index are excluded: Students who passed achieved

```r
v[v >= 7]
```

points, i.e. they achieved

```r
mean(v[v >= 7])
```

points on average.

### Multiple variables 

Let's assume the students' names are Mary, Peter, and Paul. We can store the names in a second collection:

```r
name = c("Mary", "Peter", "Paul")
```

Who passed the exam?

```r
name[v >= 7]
```

And who failed?

```r
name[v < 7]
```

### Logical Operators

Instead of "v < 7" (grade less than 7 points) we could use the exclamation mark to negate logical values (grade not greater or equal 7 points):

```r
v < 7
```

```r
!(v >= 7)
```

While 

```r
name[!(v >= 7)]
```

looks unnessecarily complex, negation comes in handy if we improve the readability of our code by computing new variables:

```r
passed = (v >= 7) # compute
passed # print to check
```

As a sidenote, see how we can add short comments directly in the R code starting with #.

These students have passed

```r
name[passed]
```

These students did not pass

```r
name[!passed]
```

Compare values with "<", "<=", ">", ">=" regarding how they differ and use "==", "!=" to check if values are equal or unequal. 

To look up Mary's grade we use the following vector to "filter": 

```r
name == "Mary"
```

which results in 

```r
v[name == "Mary"]
```

Logical values can be combined with & (*and*) and | (*or*). They can even be set by T (TRUE) and F (FALSE). That said, do you understand the following?

```r
T&T
```

```r
T|T
```

```r
T&F
```

```r
T|F
```

```r
F&F
```

```r
F|F
```

```r
T&!F
```

```r
F|!F
```

We can apply this by asking for students who passed, but did not reach the full grade (i.e., students who reached 7 points at minimum and below 12 points at maximum):

```r
name[v >= 7 & v < 12]
```

### Dataframes

Data is commonly organized as data sets, or *dataframes* in the R terminology. A simple dataframe for our data looks like this:

```r
df = data.frame(Grade = v, Name = name)
df
```

Note how R automatically assigns proper data classes to our data, such as *str* (string/text) for the names and *dbl* (double/decimal) for the grades.

We have now organized the data as a two-dimensional table with three rows and two columns, i.e., as a 3x2 table. Data are now assessed by the column names. Therefore, to access a specific variable, we write

```r
df$Grade
```

to simply print all values of the variable *Grade*, while

```r
mean(df$Grade)
```

returns the mean of all grades.

As we have two dimensions (rows and columns), specific values are indexed by two indices. The row index is defined first by convention.

```r
df[2, 1]
```

```r
df[1, 2]
```

If we leave an index undefined, the complete vector is returned. Values of all rows in second column are 

```r
df[,2]
```

Accordingly, values in the third row for all columns are

```r
df[3,]
```

Note the difference in output formats. While `df[,2]` returns a list of values (a *column vector*), `df[3,]` returns a table (a *dataframe*). In order to get a dataframe when selecting columns, simply omit the comma:

```r
df[2]
```

This difference might look weird for you at the moment, but it will become more important as you progress as an R expert. Here is another one: Columns can be accessed by their comumn name, too. Because the second column is named *Name*, `df[,2]` can be written as `df[,"Name"]`, too.

```r
df[,"Name"]
```

or, as a shortcut

```r
df$Name
```

### Filter a dataframe

In real life, data is filtered by content. To limit the dataset to students who passed, define

```r
df[df$Grade >= 7,]
```

Columns (Variables) can be filtered equivalently: Mary's grade is

```r
df$Grade[df$Name == "Mary"]
```

the average grade of all students who passed is

```r
mean(df$Grade[df$Grade >=7])
```

and the span of all grades is

```r
max(df$Grade) - min(df$Grade)
```
